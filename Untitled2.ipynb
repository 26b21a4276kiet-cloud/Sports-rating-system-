```python
# ============================================================
# INPUT DATA
# ============================================================

teams = ["Team A", "Team B", "Team C", "Team D", "Team E"]

# Each match is:
# (Winner, Loser, Point Differential)

matches = [
    ("Team A", "Team B", 5),
    ("Team B", "Team C", 3),
    ("Team C", "Team D", 4),
    ("Team D", "Team E", 2),
    ("Team E", "Team A", 1),

    ("Team A", "Team C", 2),
    ("Team B", "Team D", 1),
    ("Team C", "Team E", 3),
    ("Team D", "Team A", 2),
    ("Team E", "Team B", 4)
]

print("Teams:")
for team in teams:
    print("-", team)

print("\nNumber of matches:", len(matches))
```

```python
import numpy as np
import pandas as pd

def calculate_team_ratings(teams, matches):
    n = len(teams)
    team_index = {team: i for i, team in enumerate(teams)}

    # Create point-differential matrix
    M = np.zeros((n, n))

    for winner, loser, margin in matches:
        i = team_index[winner]
        j = team_index[loser]

        M[i][j] += margin
        M[j][i] -= margin

    # Find eigenvalues and eigenvectors
    eigenvalues, eigenvectors = np.linalg.eig(M)

    # Select eigenvector corresponding to the eigenvalue
    # with the largest absolute value
    idx = np.argmax(np.abs(eigenvalues))

    dominant_eigenvalue = eigenvalues[idx].real
    ratings = eigenvectors[:, idx].real

    # Make ratings positive for easier interpretation
    if np.sum(ratings) < 0:
        ratings = -ratings

    # Normalize ratings
    ratings = ratings / np.sum(ratings)

    # Create ranking
    ranking = pd.DataFrame({
        "Team": teams,
        "Rating": ratings
    })

    ranking = ranking.sort_values(
        by="Rating",
        ascending=False
    ).reset_index(drop=True)

    ranking["Rank"] = ranking.index + 1

    ranking = ranking[["Rank", "Team", "Rating"]]

    return M, dominant_eigenvalue, ratings, ranking


# ============================================================
# CHANGE THESE VALUES FOR A NEW EXAMPLE
# ============================================================

teams = [
    "Thunder Squad",
    "Warriors",
    "Lions",
    "Tigers",
    "Panthers"
]

matches = [
    ("Thunder Squad", "Warriors", 5),
    ("Warriors", "Lions", 3),
    ("Lions", "Tigers", 4),
    ("Tigers", "Panthers", 2),
    ("Panthers", "Thunder Squad", 1),

    ("Thunder Squad", "Lions", 2),
    ("Warriors", "Tigers", 1),
    ("Lions", "Panthers", 3),
    ("Tigers", "Thunder Squad", 2),
    ("Panthers", "Warriors", 4)
]


# ============================================================
# CALCULATE TEAM RATINGS
# ============================================================

M, dominant_eigenvalue, ratings, ranking = calculate_team_ratings(
    teams,
    matches
)


# ============================================================
# DISPLAY RESULTS
# ============================================================

print("POINT DIFFERENTIAL MATRIX:")
print(pd.DataFrame(M, index=teams, columns=teams))

print("\nDOMINANT EIGENVALUE:")
print(dominant_eigenvalue)

print("\nTEAM RATINGS:")
for team, rating in zip(teams, ratings):
    print(f"{team}: {rating:.4f}")

print("\nFINAL SPORTS TEAM RANKING")
display(ranking)
```

```python
# ============================================================
# RUN COMPLETE PROJECT
# ============================================================

M, dominant_eigenvalue, ratings, ranking = calculate_team_ratings(
    teams,
    matches
)

print("POINT DIFFERENTIAL MATRIX")
display(pd.DataFrame(M, index=teams, columns=teams))

print("\nDOMINANT EIGENVALUE")
print(f"{dominant_eigenvalue:.6f}")

print("\nTEAM RANKING")
display(ranking)
```
